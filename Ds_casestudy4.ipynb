```python
# Dictionary Application using Hash Table
# Collision Resolution: Separate Chaining

class Dictionary:
    def __init__(self, size=10):
        self.size = size
        self.table = [[] for i in range(size)]

    # Hash function
    def hash_function(self, word):
        return sum(ord(ch) for ch in word.lower()) % self.size

    # Insert word and meaning
    def insert(self, word, meaning):
        index = self.hash_function(word)
        bucket = self.table[index]

        for item in bucket:
            if item[0] == word.lower():
                item[1] = meaning
                print("Meaning updated successfully!")
                return

        bucket.append([word.lower(), meaning])
        print("Word added successfully!")

    # Search word
    def search(self, word):
        index = self.hash_function(word)
        bucket = self.table[index]

        for item in bucket:
            if item[0] == word.lower():
                print("Word:", item[0])
                print("Meaning:", item[1])
                return

        print("Word not found in dictionary!")

    # Delete word
    def delete(self, word):
        index = self.hash_function(word)
        bucket = self.table[index]

        for item in bucket:
            if item[0] == word.lower():
                bucket.remove(item)
                print("Word deleted successfully!")
                return

        print("Word not found!")

    # Display dictionary
    def display(self):
        print("\nDictionary Contents:")

        for i in range(self.size):
            print("Index", i, ":", end=" ")

            if self.table[i]:
                for item in self.table[i]:
                    print(item[0], "=", item[1], end=" | ")
            else:
                print("Empty", end="")

            print()


# Main program
d = Dictionary()

# Insert words and meanings
d.insert("Apple", "A fruit")
d.insert("Book", "A collection of written pages")
d.insert("Cat", "A domestic animal")
d.insert("Dog", "A loyal domestic animal")
d.insert("Ant", "A small insect")

# Display all words
d.display()

# Search a word
print("\nSearching a word:")
d.search("Apple")

# Delete a word
print("\nDeleting a word:")
d.delete("Cat")

# Display updated dictionary
d.display()
```
